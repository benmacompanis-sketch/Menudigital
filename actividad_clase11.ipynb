## 6. Contamos los valores de una columna
Cambiá `'nombre_columna'` por la columna que quieran analizar.

```python
conteo = df['nombre_columna'].value_counts()

print(conteo.head(10))
```

## 7. Gráfico de barras

```python
conteo.head(10).plot(kind='bar', figsize=(10, 5))

plt.title('Título del gráfico')   # Cambiá el título
plt.xlabel('Categoría')
plt.ylabel('Cantidad')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
