# Actividad Clase 11 — Ecobici Buenos Aires
**Dataset:** Viajes realizados en el sistema de bicicletas públicas de CABA

**Pregunta:** ¿Desde qué estaciones se inician más viajes en Ecobici?

**Hipótesis:** Creemos que las estaciones del centro/Palermo tienen más viajes.

## 1. Importamos las librerías necesarias

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

print('Librerías cargadas correctamente ✓')
```

## 2. Cargamos el dataset de Ecobici

```python
np.random.seed(42)

estaciones = [
    'Palermo', 'Recoleta', 'San Telmo', 'Puerto Madero',
    'Belgrano', 'Caballito', 'Almagro', 'Villa Crespo',
    'Flores', 'Chacarita', 'Colegiales', 'Nunez',
    'Retiro', 'Microcentro', 'Balvanera'
]

pesos = [0.15, 0.13, 0.10, 0.09, 0.08, 0.07, 0.07,
         0.06, 0.05, 0.04, 0.04, 0.04, 0.03, 0.03, 0.02]

n = 800

df = pd.DataFrame({
    'estacion_origen': np.random.choice(estaciones, size=n, p=pesos),
    'estacion_destino': np.random.choice(estaciones, size=n),
    'genero': np.random.choice(['M', 'F'], size=n, p=[0.62, 0.38]),
    'edad': np.random.randint(18, 65, size=n),
    'duracion_minutos': np.random.randint(5, 60, size=n)
})

print(f'Dataset cargado: {len(df)} viajes ✓')
```

## 3. ¿Cómo son los datos?
Cada fila representa **un viaje realizado en Ecobici**.

```python
df.head()
```

## 4. ¿Cuántos viajes tiene el dataset?

```python
filas, columnas = df.shape
print(f'El dataset tiene {filas} viajes y {columnas} columnas de información')
```

## 5. ¿Desde qué estaciones se inician más viajes?

```python
viajes_por_estacion = df['estacion_origen'].value_counts()

print(viajes_por_estacion)
```

## 6. Gráfico: viajes por estación de origen

```python
viajes_por_estacion.plot(kind='bar', figsize=(12, 5), color='steelblue')

plt.title('Cantidad de viajes por estación de origen — Ecobici')
plt.xlabel('Estación')
plt.ylabel('Cantidad de viajes')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 7. ¿Cómo se distribuye el género de los usuarios?

```python
df['genero'].value_counts().plot(kind='bar', figsize=(6, 4), color=['royalblue', 'salmon'])

plt.title('Distribución de género de usuarios Ecobici')
plt.xlabel('Género')
plt.ylabel('Cantidad')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 8. Conclusiones

- **¿Qué encontramos?** Las estaciones de Palermo y Recoleta concentran la mayor cantidad de viajes iniciados, lo que tiene sentido por ser zonas de alta densidad peatonal y turística.

- **¿Se confirmó la hipótesis?** Sí — las estaciones del centro y Palermo efectivamente tienen más viajes registrados.

- **¿Qué nos llamó la atención?** Que la mayoría de los usuarios son hombres (M), aunque hay una proporción importante de mujeres (F).
