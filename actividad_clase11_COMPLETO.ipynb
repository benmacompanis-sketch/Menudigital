# Actividad Clase 11 — Datos Abiertos de Buenos Aires
**Dataset:** Arbolado Público de la Ciudad de Buenos Aires

**Pregunta:** ¿En qué comunas hay más árboles registrados?

**Hipótesis:** Creemos que las comunas del centro tienen más árboles registrados.

## 1. Importamos las librerías necesarias

```python
import pandas as pd
import matplotlib.pyplot as plt

print('Librerías cargadas correctamente ✓')
```

## 2. Descargamos el dataset automáticamente
No hace falta subir ningún archivo — lo descargamos directo de internet.

```python
url = 'https://raw.githubusercontent.com/ftmanoukian/BD2026/main/4B/clase11/actividad%20notebooks/arbolado-publico-lineal.csv'

df = pd.read_csv(url)

print('Dataset descargado correctamente ✓')
```

## 3. ¿Cómo son los datos?
Cada fila representa **un árbol registrado en la vía pública** de Buenos Aires.

```python
df.head()
```

## 4. ¿Cuántos árboles tiene el dataset?

```python
filas, columnas = df.shape
print(f'El dataset tiene {filas} árboles registrados y {columnas} columnas de información')
```

## 5. ¿Cuántos árboles hay por comuna?

```python
arboles_por_comuna = df['comuna'].value_counts().sort_index()

print(arboles_por_comuna)
```

## 6. Gráfico de barras

```python
arboles_por_comuna.plot(kind='bar', figsize=(10, 5), color='green')

plt.title('Cantidad de árboles por comuna en Buenos Aires')
plt.xlabel('Comuna')
plt.ylabel('Cantidad de árboles')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 7. ¿Cuáles son las especies más comunes?

```python
especies = df['nombre_com'].value_counts().head(10)

especies.plot(kind='barh', figsize=(10, 5), color='steelblue')

plt.title('Las 10 especies de árboles más comunes')
plt.xlabel('Cantidad')
plt.tight_layout()
plt.show()
```

## 8. Conclusiones

- **¿Qué encontramos?** Algunas comunas tienen muchos más árboles registrados que otras. Las comunas con más espacio verde o calles más anchas tienden a tener más árboles.

- **¿Se confirmó la hipótesis?** No necesariamente — los datos muestran que las comunas del sur y oeste también tienen gran cantidad de árboles, no solo el centro.

- **¿Qué nos llamó la atención?** La gran variedad de especies de árboles que existen en la ciudad.
